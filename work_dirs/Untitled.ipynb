```python
import os
import json
import glob
import matplotlib.pyplot as plt

# ==========================================
# 1. 只需要填子文件夹的名字（因为你的笔记本已经在 work_dirs 里了）
folder_name = 'stream_petr_r50_flash_704_bs2_seq_24e'
# ==========================================

# 自动寻找该文件夹下最新的 .log.json 文件
try:
    # 获取所有 .log.json 文件
    list_of_files = glob.glob(os.path.join(folder_name, '*.log.json'))
    
    if not list_of_files:
        print(f"❌ 在文件夹 '{folder_name}' 里没找到 .log.json 文件！")
        print("请确认你是否已经开始了训练，并且生成了日志。")
    else:
        # 找到最新的那个文件
        latest_file = max(list_of_files, key=os.path.getctime)
        print(f"✅ 自动锁定最新日志文件: {latest_file}")

        # 开始读取数据
        iters = []
        losses = []
        
        with open(latest_file, 'r') as f:
            for line in f:
                try:
                    log = json.loads(line.strip())
                    # 只提取训练数据 (mode='train')
                    if log.get('mode') == 'train' and 'loss' in log:
                        iters.append(log['iter'])
                        losses.append(log['loss'])
                except:
                    pass

        # 画图
        if len(losses) > 0:
            plt.figure(figsize=(12, 6))
            plt.plot(iters, losses, label='Total Loss', color='red', linewidth=1.5)
            plt.xlabel('Iterations')
            plt.ylabel('Loss')
            plt.title(f'Training Loss Curve\n(File: {latest_file})')
            plt.grid(True, linestyle='--', alpha=0.7)
            plt.legend()
            plt.show()
            print(f"🚀 当前最新 Loss: {losses[-1]}")
            if len(losses) > 10:
                print(f"📉 相比 10 步之前: {'下降了' if losses[-1] < losses[-11] else '上升了'}")
        else:
            print("⚠️ 日志文件是空的，可能训练才刚刚开始，还没来得及写入数据。请稍等几分钟再运行。")

except Exception as e:
    print(f"❌ 发生未知错误: {e}")
```
